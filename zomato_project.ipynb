# Step 1 -Importing libraries

```python
# pandas is used for data manipulation and analysis.
# numpy is used for numerical operation.
# matplotlib.pyplot and seabron are used for data visualization
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Step 2- Create the data frame

```python
dataframe = pd.read_csv("/content/Zomato data .csv")
print(dataframe)
```

# Convert the data type of column - rate

```python
def handleRate(value):
        value = str(value).split('/')
        value = value[0]
        return float(value)
dataframe['rate'] = dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

```python
dataframe.info()
```

# Type of reaturant

```python
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("type of resturant")
```

```python
dataframe.head()
```

```python
grouped_data = dataframe.groupby('listed_in(type)')['votes'].sum()
result=pd.DataFrame({'votes':grouped_data})
plt.plot(result, c="blue",marker ="o")
plt.xlabel("type of resturant" , c ="red", size=20)
plt.ylabel("votes" , c ="red", size=20)
```

# conclusion-dinning resturats has recieved maximum votes

```python
dataframe.head()
```

```python
plt.hist(dataframe['rate'] , bins=5)
plt.title("ratings distribution")
plt.show()
```

# conclusion - the majority resturat received rating frome 3.5 to 4

## which mode receives maximum rating

```python
dataframe.head()
```

```python
plt.figure(figsize=(6,6))
sns.boxplot(x='online_order',y='rate',data=dataframe)
```

# conclusion - offline order received lower rating  in comparison to online order

```python
dataframe.head()
```

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap="YlGnBu", fmt='d')
plt.title("Heatmap")
plt.xlabel("online order")
plt.ylabel("listed In (Type) ")
plt.show()
```

## conclusion: Dining rstaurants primarily accept order offline orders , whereas cafes primarily receive online orders. This suggests that clients prefers orders in person ar restaurants ,but prefer online orderig at cafes
